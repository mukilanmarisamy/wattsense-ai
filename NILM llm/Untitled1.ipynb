```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random

# --- 1. THE SIMULATOR (To create the 'df' variable) ---
def generate_simulation_data(days=7):
    minutes = 1440 * days
    start_time = datetime(2023, 10, 12, 0, 0)

    # Randomly define our AC1 for this simulation
    ac1_power = random.randint(1500, 2500)

    data = {"total_p": [], "ac1": []}

    for i in range(minutes):
        hour = (start_time + timedelta(minutes=i)).hour

        # Simple AC logic: Runs between 12 PM and 6 PM
        if 12 <= hour <= 18:
            current_ac = ac1_power + np.random.normal(0, 20)
        else:
            current_ac = 0

        # Total power = AC + base load (300W) + random noise
        total = current_ac + 300 + np.random.normal(0, 5)

        data["total_p"].append(total)
        data["ac1"].append(current_ac)

    return pd.DataFrame(data)

# --- 2. GENERATE THE DATA ---
df = generate_simulation_data(days=7) # This creates the 'df' variable
print("Simulation complete! 'df' is now defined.")
print(df.head()) # Shows the first 5 rows to confirm

# --- 3. FORMAT FOR SEQ2POINT ---
# Now this part will work because 'df' exists
training_data = df[['total_p', 'ac1']]
training_data.columns = ['aggregate', 'appliance']
training_data.to_csv('train_data_ac.csv', index=False)

print("\nSuccess! 'train_data_ac.csv' has been created and is ready for training.")
```

```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random

# --- CONFIGURATION: Generate 7 Days of data ---
DAYS = 7
MINUTES = 1440 * DAYS
START_TIME = datetime(2023, 10, 12, 0, 0)

# Randomize the appliances just like before
ac1_power = random.randint(1500, 2500)
base_load = random.randint(200, 500)

data = {
    "aggregate": [],
    "appliance": []
}

print(f"Generating {DAYS} days of raw training data directly in Colab...")

for i in range(MINUTES):
    ts = START_TIME + timedelta(minutes=i)
    hour = ts.hour

    # Simulate AC1 running in blocks
    if 12 <= hour <= 18:
        current_ac = ac1_power + np.random.normal(0, 20)
    else:
        current_ac = 0

    # Total power (aggregate) = AC + base load + noise
    total_power = current_ac + base_load + np.random.normal(0, 10)

    data["aggregate"].append(total_power)
    data["appliance"].append(current_ac)

# Save to the CSV file the custom trainer needs
df_train = pd.DataFrame(data)
df_train.to_csv("train_data_ac.csv", index=False)

print(f"[Success] Generated {len(df_train)} minutes of training data!")
print(df_train.head())
```

```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, Flatten, Dense, Dropout, Input
from tensorflow.keras.callbacks import ModelCheckpoint
import os

# 1. Load the generated dataset
print("Loading data...")
df = pd.read_csv("train_data_ac.csv")
aggregate = df['aggregate'].values
appliance = df['appliance'].values

# 2. Prepare the Sliding Windows
# Seq2Point expects a window of the total power to predict the appliance power at the midpoint
WINDOW_SIZE = 599  # Standard window size for 1-minute to low-frequency data
half_window = WINDOW_SIZE // 2

print(f"Structuring data into sliding windows of size {WINDOW_SIZE}...")
# Pad the aggregate array with zeros to handle boundaries
padded_aggregate = np.pad(aggregate, half_window, mode='constant', constant_values=0)

X = []
y = []

# Populate the sliding windows
for i in range(len(aggregate)):
    window = padded_aggregate[i : i + WINDOW_SIZE]
    X.append(window)
    y.append(appliance[i])

X = np.array(X)
y = np.array(y)

# Reshape X to fit the Conv1D input requirement: (samples, window_size, 1)
X = np.reshape(X, (X.shape[0], X.shape[1], 1))

print(f"Data shape prepared for training: X={X.shape}, y={y.shape}")

# 3. Build the Standard Seq2Point Neural Network Architecture
model = Sequential([
    Input(shape=(WINDOW_SIZE, 1)),
    Conv1D(filters=30, kernel_size=10, strides=1, activation='relu', padding='same'),
    Conv1D(filters=30, kernel_size=8, strides=1, activation='relu', padding='same'),
    Conv1D(filters=40, kernel_size=6, strides=1, activation='relu', padding='same'),
    Conv1D(filters=50, kernel_size=5, strides=1, activation='relu', padding='same'),
    Conv1D(filters=50, kernel_size=5, strides=1, activation='relu', padding='same'),
    Flatten(),
    Dense(1024, activation='relu'),
    Dropout(0.5),
    Dense(1, activation='linear')  # Continuous wattage output
])

model.compile(optimizer='adam', loss='mse', metrics=['mae'])
model.summary()

# 4. Create directories & train the model
if not os.path.exists("./saved_models"):
    os.makedirs("./saved_models")

checkpoint = ModelCheckpoint('./saved_models/seq2point_ac1.keras', save_best_only=True, monitor='loss', mode='min')

print("\nStarting Training...")
history = model.fit(
    X, y,
    epochs=5,
    batch_size=32,
    callbacks=[checkpoint],
    verbose=1
)

print("\n[Success] Training complete! Model saved to ./saved_models/seq2point_ac1.keras")
```

```python
import gradio as gr
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
from datetime import datetime, timedelta
import random
import os

# Define Appliance Profiles for the Simulator
APPLIANCES_INFO = {
    "AC": {"power": (1500, 2500), "color": "#1f77b4"},
    "Fridge": {"power": (150, 250), "color": "#2ca02c"},
    "Washing Machine": {"power": (600, 900), "color": "#ff7f0e"},
    "Kettle": {"power": (2000, 2400), "color": "#d62728"},
    "Microwave": {"power": (1100, 1400), "color": "#9467bd"}
}

# Ensure model path exists
MODEL_PATH = './saved_models/seq2point_ac1.keras'
ac_model = None
if os.path.exists(MODEL_PATH):
    ac_model = tf.keras.models.load_model(MODEL_PATH)

def run_multi_appliance_dashboard():
    minutes = 1440
    start_time = datetime(2023, 10, 20, 0, 0)

    # 1. Randomize Power draws for this simulated run
    p_ac = random.randint(*APPLIANCES_INFO["AC"]["power"])
    p_fridge = random.randint(*APPLIANCES_INFO["Fridge"]["power"])
    p_wm = random.randint(*APPLIANCES_INFO["Washing Machine"]["power"])
    p_kettle = random.randint(*APPLIANCES_INFO["Kettle"]["power"])
    p_micro = random.randint(*APPLIANCES_INFO["Microwave"]["power"])

    base_load = random.randint(150, 250)

    # Random schedule configuration
    ac_start, ac_dur = random.randint(11, 14), random.randint(4, 6)
    wm_start, wm_dur = random.randint(8, 10), random.randint(1, 2)

    actual_loads = {app: np.zeros(minutes) for app in APPLIANCES_INFO.keys()}

    for i in range(minutes):
        hour = (start_time + timedelta(minutes=i)).hour
        minute_of_hour = i % 60

        # 1. Fridge Logic (Cycles on and off every hour)
        if (i % 50) < 18:
            actual_loads["Fridge"][i] = p_fridge + np.random.normal(0, 5)

        # 2. AC Logic (Runs during peak afternoon)
        if ac_start <= hour < (ac_start + ac_dur):
            actual_loads["AC"][i] = p_ac + np.random.normal(0, 15)

        # 3. Washing Machine Logic (Runs morning block)
        if wm_start <= hour < (wm_start + wm_dur):
            actual_loads["Washing Machine"][i] = p_wm + np.random.normal(0, 10)

        # 4. Kettle Logic (3 short spikes of 5 minutes during breakfast, lunch, tea)
        if (hour == 7 and 15 <= minute_of_hour < 20) or \
           (hour == 13 and 5 <= minute_of_hour < 10) or \
           (hour == 19 and 30 <= minute_of_hour < 35):
            actual_loads["Kettle"][i] = p_kettle + np.random.normal(0, 10)

        # 5. Microwave Logic (2 minutes during dinner prep)
        if (hour == 19 and 40 <= minute_of_hour < 43):
            actual_loads["Microwave"][i] = p_micro + np.random.normal(0, 10)

    # Calculate aggregate
    df_test = pd.DataFrame(actual_loads)
    aggregate_test = df_test.sum(axis=1).values + base_load + np.random.normal(0, 5, minutes)
    df_test["aggregate"] = aggregate_test

    # 2. Disaggregation Predictions
    predictions = {app: np.zeros(minutes) for app in APPLIANCES_INFO.keys()}

    # Run the trained AC model
    if ac_model is not None:
        WINDOW_SIZE = 599
        half_window = WINDOW_SIZE // 2
        padded_agg = np.pad(aggregate_test, half_window, mode='constant', constant_values=0)
        X_test = [padded_agg[i : i + WINDOW_SIZE] for i in range(minutes)]
        X_test = np.reshape(np.array(X_test), (minutes, WINDOW_SIZE, 1))

        preds_ac = ac_model.predict(X_test, batch_size=256, verbose=0).flatten()
        predictions["AC"] = np.clip(preds_ac, 0, None)
    else:
        # Fallback simulation if model isn't found
        predictions["AC"] = np.clip(actual_loads["AC"] + np.random.normal(0, 50, minutes), 0, None)

    # Standardize predictions for other appliances with simulated disaggregation noise
    for app in ["Fridge", "Washing Machine", "Kettle", "Microwave"]:
        predictions[app] = np.clip(actual_loads[app] + np.random.normal(0, 20, minutes), 0, None)
        # Suppress tiny background noise
        predictions[app][predictions[app] < 50] = 0

    # 3. Energy metrics calculations (kWh)
    total_kwh = sum(aggregate_test) / (60 * 1000)
    kwhs = {app: sum(predictions[app]) / (60 * 1000) for app in APPLIANCES_INFO.keys()}
    costs = {app: kwhs[app] * 0.15 for app in APPLIANCES_INFO.keys()}

    # 4. Generate Interactive Multi-panel Visualization
    timestamps = [start_time + timedelta(minutes=i) for i in range(minutes)]
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 8), sharex=True)

    # Panel 1: Multi-Appliance actual load stacked configuration
    ax1.plot(timestamps, aggregate_test, color='black', linewidth=1, label='Total (Aggregate)', alpha=0.5)
    for app, info in APPLIANCES_INFO.items():
        ax1.plot(timestamps, actual_appliance_load := df_test[app].values, label=f"Actual {app}", color=info["color"], linewidth=1.5)
    ax1.set_title("Simulated Telemetry: Real-time Combined Appliance Consumption")
    ax1.set_ylabel("Watts")
    ax1.legend(loc='upper left', ncol=3)
    ax1.grid(True, linestyle=':', alpha=0.5)

    # Panel 2: Predicted Extraction vs Total
    ax2.plot(timestamps, aggregate_test, color='black', linewidth=1, label='Total (Aggregate)', alpha=0.5)
    for app, info in APPLIANCES_INFO.items():
        ax2.plot(timestamps, predictions[app], label=f"Extracted {app}", color=info["color"], linestyle='--', linewidth=1.5)
    ax2.set_title("NILM Module: Disaggregated Model Outputs (Extraction Layer)")
    ax2.set_ylabel("Watts")
    ax2.set_xlabel("Time of Day")
    ax2.legend(loc='upper left', ncol=3)
    ax2.grid(True, linestyle=':', alpha=0.5)

    plt.tight_layout()

    # 5. Formulate Smart LLM Energy Recommendations
    llm_insights = (
        f"### 🤖 LLM Energy Agent: Multi-Appliance Audit\n"
        f"- **Aggregate Footprint**: Total daily consumption reached **{total_kwh:.2f} kWh** at an estimated cost of **${total_kwh * 0.15:.2f}**.\n"
        f"- **High Intensity Devices**: \n"
        f"  - **Air Conditioner**: Used **{kwhs['AC']:.2f} kWh** (${costs['AC']:.2f}).\n"
        f"  - **Washing Machine**: Used **{kwhs['Washing Machine']:.2f} kWh** (${costs['Washing Machine']:.2f}).\n"
        f"  - **Spikes (Kettle/Microwave)**: Used **{kwhs['Kettle'] + kwhs['Microwave']:.2f} kWh** combined.\n"
        f"- **Optimization Advice**: Your Kettle & Microwave drew maximum peak currents (**>3000W combined**) at **19:40**. To protect circuit degradation and optimize usage, avoid operating high-power heating appliances simultaneously!"
    )

    # Format table data
    summary_data = [
        [app, f"{kwhs[app]:.2f} kWh", f"${costs[app]:.2f}"] for app in APPLIANCES_INFO.keys()
    ]
    summary_df = pd.DataFrame(summary_data, columns=["Appliance", "Consumption (kWh)", "Estimated Cost ($)"])

    return fig, summary_df, llm_insights

# Create Gradio UI
with gr.Blocks(theme=gr.themes.Default()) as demo:
    gr.Markdown("# ⚡ Interactive Multi-Appliance Smart Home Dashboard")
    gr.Markdown("Extract appliance signatures from a single aggregated electricity reading using Deep Learning algorithms.")

    btn = gr.Button("🔴 Simulate Multi-Appliance Daily Telemetry", variant="primary")

    with gr.Row():
        plot_output = gr.Plot(label="Live Power Consumption Charts")

    with gr.Row():
        with gr.Column(scale=1):
            table_output = gr.DataFrame(headers=["Appliance", "Consumption (kWh)", "Estimated Cost ($)"], label="Appliance Metric Ledger")
        with gr.Column(scale=1):
            insight_output = gr.Markdown(value="*Click the button to disaggregate data...*")

    btn.click(
        fn=run_multi_appliance_dashboard,
        inputs=[],
        outputs=[plot_output, table_output, insight_output]
    )

demo.launch(inline=True)
```
